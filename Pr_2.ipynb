```python
# Creating a dictionary
student = {
    "name": "Krupa",
    "age": 20,
    "course": "BE"
}

print("Original Dictionary:", student)

# Adding a new item
student["city"] = "Surat"
print("After adding:", student)

# Updating a value
student["age"] = 21
print("After updating:", student)

# Accessing a value
print("Name:", student["name"])

# Getting all keys
print("Keys:", student.keys())

# Getting all values
print("Values:", student.values())

# Removing an item
student.pop("city")
print("After removing:", student)
```

```python
# Creating a list
numbers = [10, 20, 30, 20, 40]

print("Original List:", numbers)

# Adding an element
numbers.append(50)
print("After append:", numbers)

# Inserting an element
numbers.insert(1, 15)
print("After insert:", numbers)

# Removing an element
numbers.remove(20)
print("After remove:", numbers)

# Sorting
numbers.sort()
print("After sort:", numbers)

# Reversing
numbers.reverse()
print("After reverse:", numbers)

# Finding length
print("Length:", len(numbers))
```

```python
# Creating a tuple
numbers = (10, 20, 30, 20, 40)

print("Tuple:", numbers)

# Count an element
print("Count of 20:", numbers.count(20))

# Find position
print("Index of 30:", numbers.index(30))

# Length
print("Length:", len(numbers))
```

```python
# Creating a set
numbers = {10, 20, 30, 20, 40}

print("Original Set:", numbers)

# Adding an element
numbers.add(50)
print("After add:", numbers)

# Removing an element
numbers.remove(20)
print("After remove:", numbers)

# Creating another set
other = {30, 40, 60, 70}

# Union
print("Union:", numbers.union(other))

# Intersection
print("Intersection:", numbers.intersection(other))

# Difference
print("Difference:", numbers.difference(other))
```

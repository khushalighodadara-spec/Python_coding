```python
from scipy import stats
import numpy as np
data = [10,20,20,30,40,50,60,70,80,90]
#Statistical measures
print("Mean:",np.mean(data))
print("Median:",np.median(data))
print("Mode:",stats.mode(data,keepdims = True).mode[0])
print("Variance:",np.var(data))
print("Standard Deviation:",np.std(data))
```

```python
#statistical tests
#1.Normality test
result = stats.shapiro(data) # Define 'result' using stats.shapiro()
print("shapiro-wilk Test :")
print("Statistic:", result.statistic)
print("P-value:", result.pvalue)

# 2. One-sample t-test
t_result = stats.ttest_1samp(data, 15)
print("\nOne Sample T-Test:")
print("T-statistic:", t_result.statistic)
print("P-value:", t_result.pvalue)
```

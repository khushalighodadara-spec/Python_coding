```python

# Function with no arguments and no return value

def greet():
    print("Welcome to Python Programming!")

greet()
```

```python
# Function with arguments and no return value

def display(name, age):
    print("Name:", name)
    print("Age:", age)

display("Alice", 20)
```

```python
# Function with arguments and return value

def add(a, b):
    return a + b

result = add(25, 25)
print("Sum =", result)
```

```python
# Function with default arguments

def interest(principal, rate=5, time=2):
    si = (principal * rate * time) / 100
    return si

print("Interest =", interest(10000))
print("Interest =", interest(10000, 8, 3))
```

```python
# Function with keyword arguments

def student(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student(course="Python", name="John", age=21)
```

```python
# Function with variable-length arguments

def total(*numbers):
    print("Numbers:", numbers)
    print("Sum =", sum(numbers))

total(10, 20, 30)
total(5, 10, 15, 20, 25)
```

```python
import math

num = 25

print("Square Root:", math.sqrt(num))
print("Power:", math.pow(5, 3))
print("Ceiling:", math.ceil(4.2))
print("Floor:", math.floor(4.8))
print("Factorial:", math.factorial(5))
print("Value of Pi:", math.pi)
```

```python
import random

print("Random Integer:", random.randint(1, 100))

colors = ["Red", "Blue", "Green", "Yellow"]
print("Random Choice:", random.choice(colors))

random.shuffle(colors)
print("Shuffled List:", colors)

print("Random Float:", random.random())
```

```python
import statistics

data = [20, 40, 60, 80, 100]

print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Standard Deviation:", statistics.stdev(data))
```

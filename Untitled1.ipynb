```python
# Variable Assignment and Data Types
age = 25                  # int (Integer)
height = 5.9              # float (Floating point)
name = "Alice"            # str (String)
is_student = True         # bool (Boolean)

# Printing variables and their types
print("--- Variables & Data Types ---")
print(f"Name: {name} | Type: {type(name)}")
print(f"Age: {age} | Type: {type(age)}")
print(f"Height: {height} | Type: {type(height)}")
print(f"Is Student: {is_student} | Type: {type(is_student)}")
```

```python
# ---------------------------------------------------------
# 1. Arithmetic Operators
# ---------------------------------------------------------
a, b = 15, 4
print("\n--- Arithmetic Operators ---")
print(f"{a} + {b}  = {a + b}")    # Addition: 19
print(f"{a} - {b}  = {a - b}")    # Subtraction: 11
print(f"{a} * {b}  = {a * b}")    # Multiplication: 60
print(f"{a} / {b}  = {a / b}")    # Division (Float): 3.75
print(f"{a} // {b} = {a // b}")  # Floor Division (Int): 3
print(f"{a} % {b}  = {a % b}")    # Modulus (Remainder): 3
print(f"{a} ** {b} = {a ** b}")   # Exponentiation (15^4): 50625

# ---------------------------------------------------------
# 2. Assignment Operators
# ---------------------------------------------------------
print("\n--- Assignment Operators ---")
x = 10
print(f"Initial x: {x}")
x += 5;  print(f"x += 5  -> {x}")
x -= 3;  print(f"x -= 3  -> {x}")
x *= 2;  print(f"x *= 2  -> {x}")
x /= 4;  print(f"x /= 4  -> {x}")
x %= 4;  print(f"x %= 4  -> {x}")
x **= 3; print(f"x **= 3 -> {x}")

# ---------------------------------------------------------
# 3. Comparison (Relational) Operators
# ---------------------------------------------------------
x, y = 10, 20
print("\n--- Comparison Operators ---")
print(f"{x} == {y} : {x == y}")   # Equal
print(f"{x} != {y} : {x != y}")   # Not equal
print(f"{x} > {y}  : {x > y}")    # Greater than
print(f"{x} < {y}  : {x < y}")    # Less than
print(f"{x} >= {y} : {x >= y}")   # Greater than or equal to
print(f"{x} <= {y} : {x <= y}")   # Less than or equal to

# ---------------------------------------------------------
# 4. Logical Operators
# ---------------------------------------------------------
p, q = True, False
print("\n--- Logical Operators ---")
print(f"{p} and {q} : {p and q}") # True if both are true
print(f"{p} or {q}  : {p or q}")  # True if at least one is true
print(f"not {p}     : {not p}")    # Inverts boolean state

# ---------------------------------------------------------
# 5. Membership Operators
# ---------------------------------------------------------
text = "Python Programming"
numbers = [1, 2, 3, 4, 5]
print("\n--- Membership Operators ---")
print(f"'Python' in text      : {'Python' in text}")
print(f"'Java' not in text     : {'Java' not in text}")
print(f"3 in numbers           : {3 in numbers}")
print(f"10 not in numbers      : {10 not in numbers}")

# ---------------------------------------------------------
# 6. Identity Operators
# ---------------------------------------------------------
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print("\n--- Identity Operators ---")
# 'is' checks if both variables refer to the EXACT same object in memory
print(f"list1 is list3     : {list1 is list3}")      # True (same memory reference)
print(f"list1 is list2     : {list1 is list2}")      # False (equal values, but different objects)
print(f"list1 is not list2 : {list1 is not list2}")  # True

# ---------------------------------------------------------
# 7. Bitwise Operators (Operates on binary bits)
# ---------------------------------------------------------
# 5 in binary  = 0101
# 3 in binary  = 0011
m, n = 5, 3
print("\n--- Bitwise Operators ---")
print(f"{m} & {n}  (AND) : {m & n}")   # 0101 & 0011 = 0001 (1)
print(f"{m} | {n}  (OR)  : {m | n}")   # 0101 | 0011 = 0111 (7)
print(f"{m} ^ {n}  (XOR) : {m ^ n}")   # 0101 ^ 0011 = 0110 (6)
print(f"~{m}     (NOT) : {~m}")      # ~0101 = -6 (2's complement)
print(f"{m} << 1 (Left Shift)  : {m << 1}")  # 0101 << 1 = 1010 (10)
print(f"{m} >> 1 (Right Shift) : {m >> 1}")  # 0101 >> 1 = 0010 (2)
```

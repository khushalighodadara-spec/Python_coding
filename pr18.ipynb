```python
import pandas as pd
import numpy as np

# Create dataset
data = {
    "Name": ["A", "B", "C", "D", "D"],
    "Age": [20, 21, np.nan, 22, 22],
    "Marks": [80, 90, 85, 500, 500]
}

df = pd.DataFrame(data)

print("========== ORIGINAL DATA ==========")
print(df)

# 1. Detect missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 2. Handle missing value
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\n========== AFTER HANDLING MISSING VALUES ==========")
print(df)

# 3. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 4. Detect outliers using IQR
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("\n========== OUTLIER LIMITS ==========")
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Limit =", lower_limit)
print("Upper Limit =", upper_limit)

# Find outliers
outliers = df[
    (df["Marks"] < lower_limit) |
    (df["Marks"] > upper_limit)
]

print("\n========== OUTLIERS ==========")
print(outliers)

# 5. Replace outliers with median
median = df["Marks"].median()

df["Marks"] = df["Marks"].apply(
    lambda x: median if x < lower_limit or x > upper_limit else x
)

print("\n========== FINAL CLEANED DATA ==========")
print(df)
```
